# Parámetros configurables 

| **Parámetro**       | **Tipo**      | **Descripción**                                                                                     | **Ejemplo**                                      |
|----------------------|--------------|-----------------------------------------------------------------------------------------------------|--------------------------------------------------|
| `model`             | `string`     | Nombre exacto del modelo Claude (incluye versión y fecha).                                         | `"claude-sonnet-4-5-20250929"`                  |
| `messages`          | `list`       | Lista de mensajes en formato `{role, content}`. Roles: `"user"`, `"assistant"`, `"system"`.       | `[{"role": "user", "content": "¿Quién fue Ada?"}]` |
| `max_tokens`        | `integer`    | Número máximo de tokens que puede generar el modelo en la respuesta.                               | `max_tokens=300`                                 |
| `temperature`       | `float`      | Controla la aleatoriedad (0.0 = determinista, 1.0 = más creativo).                                | `temperature=0.7`                                |
| `top_p`             | `float`      | Nucleus sampling: probabilidad acumulada para limitar tokens (similar a temperature).             | `top_p=0.9`                                      |
| `stop_sequences`    | `list`       | Lista de secuencias donde el modelo debe detener la generación.                                   | `stop_sequences=["\n\n"]`                        |
| `system`            | `string`     | Instrucciones globales para el asistente (tono, estilo, reglas).                                  | `system="Responde solo en JSON válido."`         |
| `tools`             | `list`       | Define herramientas que el modelo puede invocar (para tool calls).                                | `[{"name": "mi_tool", "input_schema": {...}}]`   |
| `tool_choice`       | `string/dict`| Forzar uso de una herramienta específica o permitir elección automática.                          | `tool_choice={"type": "tool", "name": "mi_tool"}`|
| `metadata`          | `dict`       | Información opcional para trazabilidad (no afecta la respuesta).                                  | `metadata={"id": "ejercicio-2"}`                 |
| `stream`            | `boolean`    | Si `True`, activa streaming para recibir tokens en tiempo real.                                   | `stream=True`                                    |

# Ejercicio 1: Generación simple de texto

## Objetivo
Familiarizarse con la generación básica de texto usando **Claude**. Formularemos una pregunta sencilla de conocimiento general para que el modelo responda con información.

En este caso, preguntaremos:

> **¿Quién fue Alan Turing?**

Se trata de una consulta directa de tipo biográfico. Esperamos que Claude genere unos párrafos resumiendo quién fue Alan Turing, su importancia histórica en computación, etc.  
Este ejercicio demuestra la capacidad de Claude para responder preguntas factuales abiertas en lenguaje natural.

---

### **Antes de ejecutar**
- Asegúrate de que la **API Key** esté configurada en la variable de entorno `ANTHROPIC_API_KEY`.
- El código creará un cliente de Anthropic, enviará la pregunta como mensaje de usuario y mostrará la respuesta del asistente.

---





````python
# Inicializamos el cliente.
# Nota: El cliente tomará automáticamente la API Key desde la variable de entorno ANTHROPIC_API_KEY.
# Si no está configurada, el código lanzará un error.
client = Anthropic(api_key="**************")

# Definimos el modelo a utilizar y la pregunta que queremos enviar.
modelo = "claude-3-haiku-20240307"
pregunta = "¿Quién fue Alan Turing?"

# Realizamos la llamada al modelo usando el método messages.create.
# Este método simula una conversación estilo chat, donde enviamos mensajes con roles.
# Parámetros importantes:
# - model: nombre del modelo
# - max_tokens: límite máximo de tokens en la respuesta (controla la longitud)
# - temperature: controla la aleatoriedad (1.0 = más variación, 0.0 = respuesta más determinista)
# - messages: lista de mensajes, en este caso solo uno con rol "user"
respuesta = client.messages.create(
    model=modelo,
    max_tokens=300,           # Respuesta relativamente corta (puedes ajustar según necesidad)
    temperature=1.0,          # Temperatura alta para respuestas más naturales y variadas
    messages=[
        {"role": "user", "content": pregunta} 
    ]
)

# La respuesta del modelo se devuelve como un objeto complejo.
# El contenido generado está en respuesta.content (lista de bloques).
# Para imprimir el texto, accedemos al primer bloque y su propiedad 'text'.
print(respuesta.content[0].text)
````

## Objetivo

1. Proponga los datos extraídos.
2. Llame a una **herramienta de validación** que:
   - Verifique la presencia de claves.

**Beneficios de los tool calls**:
- Separan **extracción** (modelo) de **validación/normalización** (código).
